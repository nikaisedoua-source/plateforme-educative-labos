## 2. Variables

Une variable garde une valeur. Calcule l'aire d'un disque de rayon $r = 5$ : $\mathcal{A} = \pi r^2$.

```python
from math import pi
r = 5
aire = pi * r ** 2
print("Aire :", round(aire, 2))
```

## 3. Boucles

## 4. Fonctions

La fonction $f(x) = x^2 - 3x + 2$ s'annule-t-elle en $x = 1$ et $x = 2$ ?

```python
def f(x):
    return x ** 2 - 3 * x + 2

for x in [0, 1, 2, 3]:
    print("f(", x, ") =", f(x))
```
